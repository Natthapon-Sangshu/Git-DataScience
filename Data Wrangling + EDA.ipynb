## เปลี่ยน Type object เป็น category สำหรับ Object ที่ควรเปลี่ยน

```python
import pandas as pd
coviddata = pd.read_csv(r'D:\รวมไฟล์ Data การเรียน\covidtotals.csv')
## coviddata = coviddata.set_index(['iso_code'])        กรณีอยากให้ iso_code เป็น index
## coviddata.columns
## coviddata.shape
## coviddata.info()
## coviddata.index.nunique() ค่า index ที่ไม่ซ้ำ
## coviddata['location'].unique()   # แสดงประเทศไม่ซ้ำ

## เปลี่ยน Type ที่เป็น object ทั้ง dataframe เปลี่ยนเป็น category เพื่อประมวลผลเร็วขึ้น
## ใน dataframe นี้ มี Object อยู่ 2 column 'lastdate', 'location'

obj_cols = coviddata.select_dtypes('object').columns  ## ได้ 'lastdate', 'location'
obj_cols
```

```python
for col in obj_cols:                             # col in obj_cols จะวิ่งตามตัวแปร obj_cols column 'lastdate', 'location'
    unique_count = coviddata[col].nunique()      # วิ่งนับจำนวนที่ไม่ซ้ำ ใน last date และ location
    total_count = len(coviddata[col])            # จะวิ่งนับจำนวนทั้งหมด  coviddata[col] ซึ่งมันจะวิ่งในตัวแปร obj_cols อีกที
    print(col, unique_count, "/", total_count)

    # ถ้า unique values น้อยกว่า 50 และซ้ำเยอะ → แปลงเป็น category
    if unique_count < 50:
        coviddata[col] = coviddata[col].astype('category')
        
coviddata.info()                                 # จากกรณีนี้ location มันมีจำนวนหลายหลายเกินจึงเป็น object เหมือนเดิม
```
