## เริ่มใช้ Dataframe นำเข้าเริ่มจาก List Dic

```python
## นำ list เข้า dataframe เป็นได้ทั้งแนว Row และ Column
## แนว Column [  ,    ,    ,    ,]

## 1. นำ Data เข้า Dataframe pd.DataFrame(data,columns = cols)

import pandas as pd
data = [10,20,30,40,'SomO']
## df = pd.DataFrame(data,columns = ['Age'])
cols = ['Age']
df = pd.DataFrame(data,columns = cols)
df
```

```python
## นำ list เข้า dataframe 2  
## แนว row [[row0  ,row0, row0],[row1  ,row1, row1],[row2  ,row2, row2]]

import pandas as pd
data3 = [['คีย์บอร์ด','อุปกรณ์คอม',1200],['ตุ๊กตา','ของเล่น',1300],['Sumsung S20','มือถือ',40000]]
cols = ['สินค้า','ประเภท','ราคา']
df3 = pd.DataFrame(data3,columns = cols)
df3
```

```python
## นำ list เข้า dataframe 3  
import pandas as pd

Product1 = ['คีย์บอร์ด','ตุ๊กตา','Sumsung S20']
Category1 = ['อุปกรณ์คอม','ของเล่น','มือถือ']
Price1 = [1200,1300,40000]

## คำสั่ง Zip จะรวม product category price =1 row ไปเรื่อยๆ จะเป็น object อ่านไม่ออก 
## ต้องครอบด้วย list จะเป็น list ของ tuple 
data4 =list(zip(Product1 ,Category1,Price1))
## data 4

cols = ['สินค้า','ประเภท','ราคา']
df4 = pd.DataFrame(data4,columns = cols)
df4
```

```python
##  นำ Dic เข้า dataframe 1  
## list ครอบด้วย Dic [{ index : data, index : data },{ index : data, index : data }]
import pandas as pd

data =[{ 'productname' :'sumsung s20','category' :'มือถือ', 'price' : 50000},{'productname' : 'ส้มโอ', 'category' : 'ผลไม้', 'price' : 150},{'productname': 'พี่ปู', 'category' : 'อาหารทะเล', 'price' : 300}]

df1 = pd.DataFrame(data)
df1

```
